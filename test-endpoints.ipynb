```python

import io
import requests
from PIL import Image


url = 'http://localhost:8080'
path = '/Users/roshantiwari/src/DJI_001/DJI_20260629172522_0001_D.JPG'
```

```python
im = Image.open(path).convert("RGB")
im.thumbnail((640, 640))
buf = io.BytesIO()
im.save(buf, format="JPEG", quality=80)

r = requests.post(
    f"{url}/v1/detect",
    files={"image": ("image.jpg", buf.getvalue(), "image/jpeg")},
    data={"seq": 1, "min_confidence": 0.4},
    timeout=30,
)
```

```python
r.json() #yolov8m
```
